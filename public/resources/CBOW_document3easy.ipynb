```python
from gensim.models import Word2Vec
import re
```

```python
with open("CBOW/document3.txt", "r") as file:
    text = file.read()

print(text)
```

```python
words = text.lower().split()
words = re.sub(r'[^a-z\s]', '', text.lower()).split()
print("Total words:", len(words))
print("Vocabulary size:", len(set(words)))
print(words)
```

```python
cbow = Word2Vec(
    sentences=[words],
    vector_size=100,
    window=2,
    min_count=1,
    sg=0,
    epochs=100
)
```

```python
print("Vocabulary Size:", len(cbow.wv))
print("Embedding Dimensions:", cbow.wv.vector_size)
```

```python
word = "virus"

print("Word:", word)
print("Embedding:", cbow.wv[word])
```

```python
display(cbow.wv.most_similar("virus", topn=15))
```

```python
print(cbow.wv.similarity("covid", "influenza"))
```
