```python
import pandas as pd
import numpy as np
import re

from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity
```

```python
df = pd.read_csv("CBOW/twitter_dataset.csv")

print(df.head())
print(df.shape)
```

```python
tweets = [
    re.sub(r'[^a-z\s]', '', tweet.lower()).split()
    for tweet in df["Text"]
]

print(tweets[0])
```

```python
cbow = Word2Vec(
    sentences=tweets,
    vector_size=100,
    window=5,
    min_count=2,
    sg=0,
    epochs=10
)

print("Vocabulary size:", len(cbow.wv))
```

```python
tweet_embeddings = np.array([
    cbow.wv.get_mean_vector([w for w in tweet if w in cbow.wv])
    for tweet in tweets
])

print("Embedding shape:", tweet_embeddings.shape)
```

```python
user_input = input("Enter a topic or keywords: ")

query_words = re.sub(r'[^a-z\s]', '', user_input.lower()).split()

query_words = [w for w in query_words if w in cbow.wv]

if len(query_words) == 0:
    print("No matching words found. Try another keyword.")
else:
    query_embedding = cbow.wv.get_mean_vector(query_words)

    similarity = cosine_similarity(
        [query_embedding], tweet_embeddings
    )[0]

    top_indices = similarity.argsort()[-5:][::-1]

    
    for index in top_indices:
        print(df["Text"].iloc[index])
        print("Similarity Score:", round(similarity[index], 3))
        print("-" * 60)
```
