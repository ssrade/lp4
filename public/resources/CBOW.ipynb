```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Embedding, Lambda
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.utils import to_categorical
```

```python
with open("cbow_text.txt", "r") as file:
    text = file.read()

tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

data = tokenizer.texts_to_sequences([text])[0]
total_vocab = len(tokenizer.word_index) + 1

print("Total words:", len(data))
print("Vocabulary size:", total_vocab)
```

```python
window_size = 2

context = []
target = []

for i in range(window_size, len(data) - window_size):
    context.append(data[i-2:i] + data[i+1:i+3])
    target.append(data[i])

context = np.array(context)
target = to_categorical(target, total_vocab)

print("Context shape:", context.shape)
print("Target shape:", target.shape)
```

```python
model = Sequential([
    Embedding(total_vocab, 100, input_length=4),
    Lambda(lambda x: tf.reduce_mean(x, axis=1)),
    Dense(total_vocab, activation="softmax")
])

model.compile(
    loss="categorical_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)

model.fit(context, target, epochs=100, verbose=1)
```

```python
print("CBOW Model Output:")

predictions = model.predict(context[:5])

for i in range(5):
    predicted_word = tokenizer.index_word[np.argmax(predictions[i])]
    actual_word = tokenizer.index_word[np.argmax(target[i])]
    print("Actual:", actual_word, "| Predicted:", predicted_word)
```
