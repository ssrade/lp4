```python
import pandas as pd
import numpy as np
import re
import matplotlib.pyplot as plt
from gensim.models import Word2Vec
```

```python
df = pd.read_csv("CBOW/twitter_dataset.csv")

print(df.head())
print(df.shape)
```

```python
tweets = [tweet.lower().split() for tweet in df["Text"]]

print(tweets[0])
```

```python
cbow = Word2Vec(
    sentences=tweets,
    vector_size=100,
    window=5,
    min_count=2,
    sg=0,
    epochs=10
)

print("Vocabulary size:", len(cbow.wv))
```

```python
word = cbow.wv.index_to_key[0]

print("Word:", word)
print("Embedding:", cbow.wv[word])
```

```python
print(cbow.wv.most_similar(word, topn=5))
```

```python
word1 = cbow.wv.index_to_key[0]
word2 = cbow.wv.index_to_key[1]

print("Similarity:", cbow.wv.similarity(word1, word2))
```
